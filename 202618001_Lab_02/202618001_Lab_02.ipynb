## Task 2 - Vectorized Arithmetic and Linear Algebra

### ● Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @ or np.matmul().

```python

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

addition = A + B

print("\nA + B:")
print(addition)

elementwise = A * B

print("\nA * B (element-wise):")
print(elementwise)

matrix_product = A @ B

print("\nA @ B (matrix multiplication):")
print(matrix_product)


```

### ● For a square matrix, find its transpose and determinant.If invertible, find the inverse and verify A @ A^-1 using np.allclose().

```python

transpose = A.T

print("\nTranspose of A:")
print(transpose)

det = np.linalg.det(A)

print("\nDeterminant of A:")
print(det)

if not np.isclose(det, 0):
    A_inv = np.linalg.inv(A)

    print("\nInverse of A:")
    print(A_inv)

    identity = A @ A_inv

    print("\nA @ A^-1:")
    print(identity)

    print("\nIs A @ A^-1 equal to identity matrix?")
    print(np.allclose(identity, np.eye(A.shape[0])))
else:
    print("\nA is not invertible.")
```

## Task 3 - Normal Distribution and Histogram

### ● Generate at least 1,000 values from a normal distribution. State the chosen mean and standard deviation.Find the sample mean and sample standard deviation. Compare them with the chosen values. Plot a histogram with it.

```python

chosen_mean = 50
chosen_std = 15

data = np.random.normal(loc=chosen_mean, scale=chosen_std, size=1000)

sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)

print(f"Chosen Mean: {chosen_mean} | Sample Mean: {sample_mean:.4f} | Diff: {abs(chosen_mean - sample_mean):.4f}")
print(f"Chosen Std: {chosen_std} | Sample Std: {sample_std:.4f} | Diff: {abs(chosen_std - sample_std):.4f}")

plt.hist(data, bins=30, edgecolor='black')
plt.show()
```

## Task 5 - Filtering and Querying


### ● Male passengers older than 50: how many?
### ● Female first-class passengers: how many, and what percentage survived?
### ● Age 20-40, Fare above overall median, and survived: how many?
### ● Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive: how many?
### ● Embarked='S', Pclass 2 or 3, and Fare above Southampton median: how many?

```python


# 1. Male passengers older than 50
q1 = df[(df['Sex'] == 'male') & (df['Age'] > 50)]
print(f"Male > 50 count: {len(q1)}")

# 2. Female first-class passengers: count and survival percentage
q2 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
print(f"Female 1st class count: {len(q2)}")
print(f"Female 1st class survival percentage: {q2['Survived'].mean() * 100:.2f}%")

# 3. Age 20-40, Fare above overall median, and survived
overall_median_fare = df['Fare'].median()
q3 = df[(df['Age'].between(20, 40)) & (df['Fare'] > overall_median_fare) & (df['Survived'] == 1)]
print(f"Age 20-40, High Fare, Survived count: {len(q3)}")

# 4. Travelling alone, age < 30, and did not survive
q4 = df[(df['SibSp'] == 0) & (df['Parch'] == 0) & (df['Age'] < 30) & (df['Survived'] == 0)]
print(f"Alone, Age < 30, Did not survive count: {len(q4)}")

# 5. Embarked='S', Pclass 2 or 3, and Fare above Southampton median
s_median_fare = df[df['Embarked'] == 'S']['Fare'].median()
q5 = df[(df['Embarked'] == 'S') & (df['Pclass'].isin([2, 3])) & (df['Fare'] > s_median_fare)]
print(f"Embarked S, Pclass 2/3, Fare > S median count: {len(q5)}")
```

## Task 6 - groupby() and Aggregation

### ● Survival rate by Sex.
### ● Survival rate by Pclass.
### ● Average Age and Fare by Pclass.
### ● Passenger count and survival rate by Sex-Pclass.
### ● Passenger count, average Fare, and survival rate by Embarked.

```python
print("--- Survival rate by Sex ---")
print(df.groupby('Sex')['Survived'].mean())

print("\n--- Survival rate by Pclass ---")
print(df.groupby('Pclass')['Survived'].mean())

print("\n--- Average Age and Fare by Pclass ---")
print(df.groupby('Pclass')[['Age', 'Fare']].mean())

print("\n--- Passenger count and survival rate by Sex-Pclass ---")
print(df.groupby(['Sex', 'Pclass'])['Survived'].agg(['count', 'mean']))

print("\n--- Passenger count, average Fare, and survival rate by Embarked ---")
print(df.groupby('Embarked').agg(
    Count=('PassengerId', 'count'),
    Avg_Fare=('Fare', 'mean'),
    Survival_Rate=('Survived', 'mean')
))
```

## Task 7 - Missing Values and Fare Outliers

### ● Find the missing-value count and percentage for every column. Plot missing counts as a bar chart.

```python
print("--- Missing Values Count and Percentage ---")
missing_count = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({'Count': missing_count, 'Percentage (%)': missing_pct})
print(missing_df)

plt.figure(figsize=(10, 5))
missing_count[missing_count > 0].plot(kind='bar', color='coral', edgecolor='black')
plt.title('Count of Missing Values per Column')
plt.ylabel('Missing Values Count')
plt.show()
```

### ● Fill missing Age values with the mean Age. Show missing counts before and after.

```python
print("\n--- Missing Age Count BEFORE Imputation ---")
print(df['Age'].isnull().sum())

df['Age'] = df['Age'].fillna(df['Age'].mean())

print("\n--- Missing Age Count AFTER Imputation ---")
print(df['Age'].isnull().sum())
```

### ● Try to fill NaN values by imputing using: mean, median, mode, random value

```python
df_mean = df.copy()
mean_val = df_mean['Age'].mean()
df_mean['Age'] = df_mean['Age'].fillna(mean_val)
print(f"Mean used: {mean_val:.2f} | NaNs after: {df_mean['Age'].isnull().sum()}")
```

```python
df_median = df.copy()
median_val = df_median['Age'].median()
df_median['Age'] = df_median['Age'].fillna(median_val)
print(f"Median used: {median_val:.2f} | NaNs after: {df_median['Age'].isnull().sum()}")
```

```python
df_mode = df.copy()
mode_val = df_mode['Age'].mode()[0]
df_mode['Age'] = df_mode['Age'].fillna(mode_val)
print(f"Mode used: {mode_val:.2f} | NaNs after: {df_mode['Age'].isnull().sum()}")
```

```python

df_random = df.copy()
nan_mask = df_random['Age'].isnull()
df_random.loc[nan_mask, 'Age'] = np.random.choice(df_random['Age'].dropna(), size=nan_mask.sum())
print(f"Random values used | NaNs after: {df_random['Age'].isnull().sum()}")
```

### ● For Fare, calculate Q1, Q3, IQR, 1.5xIQR bounds, and count the outliers.

```python
print("\n--- Fare Outliers Calculation ---")
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]
outliers_count = len(outliers)

print(f"Q1: {Q1:.4f}")
print(f"Q3: {Q3:.4f}")
print(f"IQR: {IQR:.4f}")
print(f"Lower Bound: {lower_bound:.4f}")
print(f"Upper Bound: {upper_bound:.4f}")
print(f"Total Fare Outliers: {outliers_count}")
```

### ● Create a pivot table: rows=Sex, columns=Pclass, values=mean Survived. Find the highest and lowest groups.

```python
pivot_table = df.pivot_table(values='Survived', index='Sex', columns='Pclass', aggfunc='mean')
print(pivot_table)

highest_group = pivot_table.max().max()
lowest_group = pivot_table.min().min()

print(f"\nHighest Survival Rate: {highest_group:.4f}")
print(f"Lowest Survival Rate: {lowest_group:.4f}")
```
