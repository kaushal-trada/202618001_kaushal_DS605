## Task 2 - Vectorized Arithmetic and Linear Algebra

```python

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

addition = A + B

print("\nA + B:")
print(addition)

elementwise = A * B

print("\nA * B (element-wise):")
print(elementwise)

matrix_product = A @ B

print("\nA @ B (matrix multiplication):")
print(matrix_product)


```

```python

transpose = A.T

print("\nTranspose of A:")
print(transpose)

det = np.linalg.det(A)

print("\nDeterminant of A:")
print(det)

if not np.isclose(det, 0):
    A_inv = np.linalg.inv(A)

    print("\nInverse of A:")
    print(A_inv)

    identity = A @ A_inv

    print("\nA @ A^-1:")
    print(identity)

    print("\nIs A @ A^-1 equal to identity matrix?")
    print(np.allclose(identity, np.eye(A.shape[0])))
else:
    print("\nA is not invertible.")
```
