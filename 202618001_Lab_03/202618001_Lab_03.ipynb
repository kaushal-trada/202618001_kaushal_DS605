Check Outliers Using the IQR Method

```python
outlier_results = []

for col in numerical_cols:
    
    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outlier_count = (
        (X[col] < lower_bound) |
        (X[col] > upper_bound)
    ).sum()
    
    outlier_percentage = round(
        (outlier_count / len(X)) * 100, 2
    )
    
    outlier_results.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier Percentage": outlier_percentage
    })

# Create DataFrame
outlier_df = pd.DataFrame(outlier_results)

print(outlier_df)
```


 REMOVE ONLY CLEAR / EXTREME OUTLIERS


```python
import pandas as pd


# Recreate numerical columns from current X
numerical_cols = X.select_dtypes(include="number").columns.tolist()

# Store original number of rows
original_rows = len(X)

# Create a mask: False means keep the row
extreme_outlier_mask = pd.Series(False, index=X.index)

# Store outlier information
outlier_summary = []

# Check all numerical columns
for col in numerical_cols:
    
    # Calculate Q1 and Q3
    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)
    
    # Calculate IQR
    IQR = Q3 - Q1
    
    # Use 3 × IQR to identify only extreme outliers
    lower_bound = Q1 - 3 * IQR
    upper_bound = Q3 + 3 * IQR
    
    # Identify extreme outliers
    col_outliers = (
        (X[col] < lower_bound) |
        (X[col] > upper_bound)
    )
    
    # Add to overall mask
    extreme_outlier_mask |= col_outliers
    
    # Save results
    outlier_summary.append({
        "Column": col,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Extreme Outliers": col_outliers.sum()
    })


# ============================================
# OUTLIER SUMMARY
# ============================================

outlier_summary_df = pd.DataFrame(outlier_summary)

print("Extreme Outliers in Each Numerical Column:")
display(outlier_summary_df)


# ============================================
# REMOVE EXTREME OUTLIER ROWS
# ============================================

X_clean = X.loc[~extreme_outlier_mask].copy()
y_clean = y.loc[~extreme_outlier_mask].copy()


# ============================================
# REPORT RESULTS
# ============================================

remaining_rows = len(X_clean)
removed_rows = original_rows - remaining_rows
removed_percentage = (removed_rows / original_rows) * 100

print("\n----- OUTLIER REMOVAL RESULTS -----")
print("Original number of rows:", original_rows)
print("Rows after removing extreme outliers:", remaining_rows)
print("Number of rows removed:", removed_rows)
print("Percentage of rows removed:", round(removed_percentage, 2), "%")
```

## Task 3 - Create Two Preprocessing Pipelines

```python


from sklearn.model_selection import train_test_split

# Use the dataset after outlier removal
X_final = X_clean.copy()
y_final = y_clean.copy()

# Split data: 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X_final,
    y_final,
    test_size=0.2,
    stratify=y_final,
    random_state=42
)

# ============================================
# CHECK DATASET SHAPES
# ============================================

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

# ============================================
# CHECK CLASS DISTRIBUTION
# ============================================

print("\nOriginal Class Distribution:")
print(y_final.value_counts(normalize=True))

print("\nTraining Class Distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting Class Distribution:")
print(y_test.value_counts(normalize=True))
```

use KNNImputer(n_neighbors=5) for the numerical columns.

```python
from sklearn.impute import KNNImputer

from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer

# Numerical preprocessing pipeline
numerical_pipeline = Pipeline(steps=[
    ("imputer", KNNImputer(n_neighbors=5))
])

numerical_pipeline
```

 Categorical preprocessing pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

categorical_pipeline
```

● Pipeline A: KNNImputer + StandardScaler for numerical features.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler

# Pipeline A for numerical features
numerical_pipeline_A = Pipeline(steps=[
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", StandardScaler())
])

numerical_pipeline_A
```

● Pipeline B: KNNImputer + MinMaxScaler for numerical features.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer
from sklearn.preprocessing import MinMaxScaler

# Pipeline B for numerical features
numerical_pipeline_B = Pipeline(steps=[
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", MinMaxScaler())
])

numerical_pipeline_B
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# ============================================
# IDENTIFY NUMERICAL AND CATEGORICAL COLUMNS
# ============================================

numerical_cols = X_train.select_dtypes(
    include="number"
).columns.tolist()

categorical_cols = X_train.select_dtypes(
    include="object"
).columns.tolist()


# ============================================
# PREPROCESSING PIPELINE A
# KNNImputer + StandardScaler
# ============================================

preprocessor_A = ColumnTransformer(
    transformers=[
        ("num", numerical_pipeline_A, numerical_cols),
        ("cat", categorical_pipeline, categorical_cols)
    ]
)


# ============================================
# PREPROCESSING PIPELINE B
# KNNImputer + MinMaxScaler
# ============================================

preprocessor_B = ColumnTransformer(
    transformers=[
        ("num", numerical_pipeline_B, numerical_cols),
        ("cat", categorical_pipeline, categorical_cols)
    ]
)


# ============================================
# CHECK THE PIPELINES
# ============================================

print("Preprocessing Pipeline A:")
print(preprocessor_A)

print("\nPreprocessing Pipeline B:")
print(preprocessor_B)
```

### Task 4 - Train Two Classification Models

Train Logistic Regression Using Pipeline A and Pipeline B

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# ============================================
# LOGISTIC REGRESSION + PIPELINE A
# ============================================

logistic_pipeline_A = Pipeline(steps=[
    ("preprocessor", preprocessor_A),
    ("model", LogisticRegression(max_iter=1000))
])

# Train the model
logistic_pipeline_A.fit(X_train, y_train)

logistic_pipeline_A

```

```python
# ============================================
# LOGISTIC REGRESSION + PIPELINE B
# ============================================

logistic_pipeline_B = Pipeline(steps=[
    ("preprocessor", preprocessor_B),
    ("model", LogisticRegression(max_iter=1000))
])

# Train the model
logistic_pipeline_B.fit(X_train, y_train)

logistic_pipeline_B
```

## Task 5 - Evaluate and Compare the Four Results

EVALUATE A MODEL

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd


def evaluate_model(model, X_train, y_train, X_test, y_test, model_name):
    
    # Predictions on training data
    y_train_pred = model.predict(X_train)
    
    # Predictions on testing data
    y_test_pred = model.predict(X_test)
    
    # Calculate metrics
    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_test_pred)
    
    precision = precision_score(y_test, y_test_pred)
    recall = recall_score(y_test, y_test_pred)
    f1 = f1_score(y_test, y_test_pred)
    
    # Return results
    return {
        "Experiment": model_name,
        "Training Accuracy": train_accuracy,
        "Testing Accuracy": test_accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    }


# ============================================
# EVALUATE ALL FOUR MODELS
# ============================================

results = []

# 1. Logistic Regression + Pipeline A
results.append(
    evaluate_model(
        logistic_pipeline_A,
        X_train, y_train,
        X_test, y_test,
        "Logistic Regression + Pipeline A"
    )
)

# 2. Logistic Regression + Pipeline B
results.append(
    evaluate_model(
        logistic_pipeline_B,
        X_train, y_train,
        X_test, y_test,
        "Logistic Regression + Pipeline B"
    )
)

# 3. Decision Tree + Pipeline A
results.append(
    evaluate_model(
        decision_tree_pipeline_A,
        X_train, y_train,
        X_test, y_test,
        "Decision Tree + Pipeline A"
    )
)

# 4. Decision Tree + Pipeline B
results.append(
    evaluate_model(
        decision_tree_pipeline_B,
        X_train, y_train,
        X_test, y_test,
        "Decision Tree + Pipeline B"
    )
)


# ============================================
# CREATE FINAL COMPARISON TABLE
# ============================================

results_df = pd.DataFrame(results)

# Convert values to percentages for easier reading
results_df["Training Accuracy"] = (
    results_df["Training Accuracy"] * 100
).round(2)

results_df["Testing Accuracy"] = (
    results_df["Testing Accuracy"] * 100
).round(2)

results_df["Precision"] = (
    results_df["Precision"] * 100
).round(2)

results_df["Recall"] = (
    results_df["Recall"] * 100
).round(2)

results_df["F1-Score"] = (
    results_df["F1-Score"] * 100
).round(2)


# Display final comparison table
results_df
```

```python
# ============================================
# FINAL COMPARISON TABLE
# ============================================

comparison_table = pd.DataFrame(results)

# Convert metrics to percentage
comparison_table["Training Accuracy"] = (
    comparison_table["Training Accuracy"] * 100
).round(2)

comparison_table["Testing Accuracy"] = (
    comparison_table["Testing Accuracy"] * 100
).round(2)

comparison_table["Precision"] = (
    comparison_table["Precision"] * 100
).round(2)

comparison_table["Recall"] = (
    comparison_table["Recall"] * 100
).round(2)

comparison_table["F1-Score"] = (
    comparison_table["F1-Score"] * 100
).round(2)

# Display comparison table
print("FINAL MODEL COMPARISON TABLE")
display(comparison_table)
```

 SELECT BEST MODEL FOR EACH ALGORITHM

```python

# Find best Logistic Regression result
best_lr_name = comparison_table[
    comparison_table["Experiment"].str.contains("Logistic Regression")
].sort_values(
    by="Testing Accuracy",
    ascending=False
).iloc[0]["Experiment"]

# Find best Decision Tree result
best_dt_name = comparison_table[
    comparison_table["Experiment"].str.contains("Decision Tree")
].sort_values(
    by="Testing Accuracy",
    ascending=False
).iloc[0]["Experiment"]

print("Best Logistic Regression:", best_lr_name)
print("Best Decision Tree:", best_dt_name)
```

```python
models = {
    "Logistic Regression + Pipeline A": logistic_pipeline_A,
    "Logistic Regression + Pipeline B": logistic_pipeline_B,
    "Decision Tree + Pipeline A": decision_tree_pipeline_A,
    "Decision Tree + Pipeline B": decision_tree_pipeline_B
}

best_lr_model = models[best_lr_name]
best_dt_model = models[best_dt_name]
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# ============================================
# BEST LOGISTIC REGRESSION CONFUSION MATRIX
# ============================================

plt.figure(figsize=(6, 5))

ConfusionMatrixDisplay.from_estimator(
    best_lr_model,
    X_test,
    y_test,
    display_labels=["Not Canceled", "Canceled"],
    cmap="viridis"
)

plt.title(f"Confusion Matrix - {best_lr_name}")
plt.show()


# ============================================
# BEST DECISION TREE CONFUSION MATRIX
# ============================================

plt.figure(figsize=(6, 5))

ConfusionMatrixDisplay.from_estimator(
    best_dt_model,
    X_test,
    y_test,
    display_labels=["Not Canceled", "Canceled"],
    cmap="viridis"
)

plt.title(f"Confusion Matrix - {best_dt_name}")
plt.show()
```
